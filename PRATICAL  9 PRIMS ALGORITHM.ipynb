```python

def prim(graph, n):
    visited = [False] * n
    visited[0] = True

    edges = []
    total_cost = 0

    for _ in range(n - 1):
        min_cost = float('inf')
        u = -1
        v = -1

        for i in range(n):
            if visited[i]:
                for j in range(n):
                    if not visited[j] and graph[i][j] != 0:
                        if graph[i][j] < min_cost:
                            min_cost = graph[i][j]
                            u = i
                            v = j

        if u == -1:
            print("Graph is not connected!")
            return

        edges.append((u, v, min_cost))
        total_cost += min_cost
        visited[v] = True

    print("\nMinimum Spanning Tree:")
    for u, v, cost in edges:
        print(u, "-", v, ":", cost)

    print("Total Minimum Cost:", total_cost)


# Main program
n = int(input("Enter number of vertices: "))

graph = []

print("Enter adjacency matrix:")

for i in range(n):
    row = list(map(int, input().split()))
    graph.append(row)

prim(graph, n)
```
